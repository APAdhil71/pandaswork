```python
import pandas as pd 
import matplotlib.pyplot as plt
df=pd.read_csv("C:\\Users\\adhil\\OneDrive\\Desktop\\machineworks\\pandasworks\\fraud_dectition\\Fraud Detection Transactions Dataset.csv")
```

```python
df.columns
```

```python
df.isnull().sum()
```

```python
# Plot the number of transactions by Transaction Type
count_type = df["Transaction_Type"].value_counts()
plt.bar(count_type.index, count_type.values)
plt.title("Number of Transactions by Type")
plt.xlabel("Transaction Type")
plt.ylabel("Count")
plt.show()
```

```python
# Plot Fraud vs Non-Fraud counts
fraud_count = df["Fraud_Label"].value_counts()
plt.bar(["Non-Fraud", "Fraud"], fraud_count.values)
plt.title("Fraud vs Non-Fraud Count")
plt.ylabel("Number of Transactions")
plt.show()
```

```python
# Plot distribution of Transaction Amounts
plt.hist (df["Transaction_Amount"])
plt.title("Transaction Amount Distribution")
plt.xlabel("Amount")
plt.ylabel("Frequency")
plt.show()
```

```python
# Plot count of Device Types
device_count = df["Device_Type"].value_counts()
plt.bar(device_count.index, device_count.values)
plt.title("Device Type Usage")
plt.xlabel("Device Type")
plt.ylabel("Count")
plt.show()
```

```python
# Location-wise transaction count
loc_count = df["Location"].value_counts().head()
plt.bar(loc_count.index, loc_count.values)
plt.title("Top 10 Locations by Transaction Count")
plt.ylabel("Count")
plt.show()
```
