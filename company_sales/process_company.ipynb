```python
import pandas as pd 
import matplotlib.pyplot as plt
df=pd.read_csv("C:\\Users\\adhil\\OneDrive\\Desktop\\machineworks\\pandasworks\\company_sales\\company_sales_data.csv")
```

```python
# 1. Read Total profit of all months and show it using a line plot
profit_summary=df.groupby("month_number")["total_profit"].sum()
plt.plot(profit_summary)
plt.show()
```

```python
# 2. Get total profit of all months and show line plot with the following Style properties
#  Generated line plot must include following Style properties:
#  Line Style dotted and Line-color should be red
#  Show legend at the lower right location.
#  X label name = Month Number
#  Y label name = Sold units number
#  Add a circle marker.
#  Line marker color as read
#  Line width should be 3
units_sold_summary=df.groupby("month_number")["total_units"].sum()
plt.plot(units_sold_summary.index,units_sold_summary.values,linestyle=":",marker="o",color="red")


```

```python
plt.plot(df["month_number"],df["facecream"])
plt.plot(df["month_number"],df["facewash"])
plt.plot(df["month_number"],df["toothpaste"])
plt.plot(df["month_number"],df["bathingsoap"])
plt.plot(df["month_number"],df["shampoo"])
plt.plot(df["month_number"],df["moisturizer"])


```

```python
products=["facecream","facewash","toothpaste","bathingsoap","shampoo","moisturizer"]
for p in products:
    plt.plot(df["month_number"],df[p],label=p,marker="o")
plt.legend()
```

```python
# 4.Read toothpaste sales data of each month and show it using a scatterplot
plt.scatter(df["month_number"],df["toothpaste"])
```

```python
# 5.Read face cream and facewash product sales data and show it using the bar chart with grid line
plt.bar(df["month_number"], df["facecream"])
plt.bar(df["month_number"], df["facewash"])
plt.grid()
```

```python
plt.hist(df.groupby("month_number")["total_profit"].sum(),edgecolor="black")
```

```python
# 7.Calculate total sale data for last year for each product and show it using a Pie chart
products=["facecream","facewash","toothpaste","bathingsoap","shampoo","moisturizer"]
sales_summary={p:df[p].sum()for p in products}
plt.pie(sales_summary.values(),labels=products,autopct="%1.1f%%")
```

```python
# 8. Read Bathing soap facewash of all months and display it using the Subplot
figure, axes = plt.subplots(1,2)

axes[0].plot(df["month_number"], df["bathingsoap"])
axes[1].plot(df["month_number"], df["facewash"])

axes[0].set_title("Bathing Soap Sales")
axes[1].set_title("Facewash Sales")

plt.tight_layout()
```
